 # Heart Disease Analysis
  Dataset: UCI Heart Failure Clinical Records 
  Goal: Clean and prepare data for SQL Server import, then Excel/Power BI dashboards

```python
import pandas as pd 
df = pd.read_csv("heart_failure_clinical_records_dataset.csv") 
print(df.shape) 
df.head()
```

## Check for missing values and duplicates

```python
print(df.isnull().sum()) 
print(df.duplicated().sum()) 
```

## Create age groups

```python
bins = [0, 45, 60, 75, 200] 
labels = ['31-45', '46-60', '61-75', '76+'] 
df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels)
df['age_group'].value_counts()
```

## Push cleaned data to SQL Server

```python
from sqlalchemy import create_engine
server = 'localhost\\SQLEXPRESS' 
database = 'HeartDiseaseDB'
engine = create_engine(f'mssql+pyodbc://@{server}/{database}?driver=ODBC+Driver+17+for+SQL+Server&trusted_connection=yes') 
df.to_sql('patients', con=engine, if_exists='replace', index=False) 
print('Data pushed to SQL Server successfully')
```
